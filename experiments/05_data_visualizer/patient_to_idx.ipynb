```python
import json
import glob
```

```python
patients = glob.glob("../../data/entire_yale_dataset/predictions/*")
patient_ids = [patient.split("/")[-1] for patient in patients]
len(patient_ids)
```

```python
patient_to_idx = {}
for i, patient in enumerate(patient_ids):
    patient_to_idx[patient] = i

patient_to_idx
```

```python
with open("patient_to_idx.json", "w") as f:
    json.dump(patient_to_idx, f , indent=4)
```
