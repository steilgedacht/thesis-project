```python
import os
os.chdir('/home/benjaminb/Dokumente/JKU/Semester_9/Practical_Work')
from datetime import datetime
import numpy as np
from utils.mri_dataloader import MRI_Dataloader
mdl = MRI_Dataloader()
```

```python
volume_to_radius = lambda x: (x * 3/(4*np.pi))**(1/3)

grow_rates = []
for trj in mdl.iterate_trajectories(enable_tqdm=True):
    if len(trj.allowed_dates) <= 1: continue
    sizes = trj.load_sizes(allowed_dates=True)
    delta_t = (datetime.strptime(trj.allowed_dates[-1], "%Y-%m-%d") - datetime.strptime(trj.allowed_dates[0], "%Y-%m-%d")).days
    grow_rate = (volume_to_radius(sizes[-1]) - volume_to_radius(sizes[0])) / delta_t
    grow_rates.append(grow_rate)
    print(np.mean((np.stack(grow_rates))))
```
